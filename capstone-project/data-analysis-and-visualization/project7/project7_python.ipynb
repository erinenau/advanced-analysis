```python
import pandas as pd

# ---------------------------------------
# PART 1: Load Dataset
# ---------------------------------------
df = pd.read_csv("C:/Users/erine/OneDrive/Desktop/MATHS/REAL ANALYSIS/project7/convoy_data_clean.csv")

print("Dataset loaded successfully.")
print(f"Total convoy trips: {len(df)}")
print("Columns:", list(df.columns))
print()

# ---------------------------------------
# PART 2: Given Regression Coefficients
# ---------------------------------------
b0 = 12.5      # intercept
b1 = 3.1       # weight_tons coefficient
b2 = -0.8      # speed_mph coefficient
b3 = 0.45      # distance_miles coefficient

# ---------------------------------------
# PART 3: Upcoming Convoy Inputs
# ---------------------------------------
weight_tons = 32
speed_mph = 42
distance_miles = 85

# ---------------------------------------
# PART 4: Compute Predicted Fuel Use
# ---------------------------------------
y_pred = (
    b0
    + b1 * weight_tons
    + b2 * speed_mph
    + b3 * distance_miles
)

# ---------------------------------------
# PART 5: Print Results Clearly
# ---------------------------------------
print("Upcoming Convoy Inputs:")
print(f"  Vehicle Weight: {weight_tons} tons")
print(f"  Average Speed:  {speed_mph} mph")
print(f"  Distance:       {distance_miles} miles")

print("\nPredicted Fuel Consumption:")
print(f"  {y_pred:.2f} gallons")
```
